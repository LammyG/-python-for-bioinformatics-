```python
# DNA Sequence Analyzer
# Task 1.1 - Basic Python: Validate sequence, count nucleotides, and calculate GC content -> (DNA Sequence Processing and Analysis)

# Purpose: To validate DNA sequence, count nucleotides, and calculate GC content

# Below shows the step-by-step approach in the code input/solution
# Step 1: Prompt dna_sequence as user input 
dna_sequence = input("Enter DNA sequence: ")

# Step 2: Define allowed bases
def is_valid_dna(sequence):
    """ Check if the input sequence only contains valid DNA bases: A, T, C, and G."""

# Step 3: is to validate: Define the valid characters as a set
    valid_nucleotides = set('ATGC')
    return all(nucleotide in valid_nucleotides for nucleotide in sequence)


# Step 4: Count nucleotides: Initialize a dict with keys for each base
def count_nucleotides(sequence):
    """Count the occurrences of each nucleotide in the sequence"""
    counts = {'A': 0, 'T': 0, 'G': 0, 'C': 0}
    for nucleotide in sequence:
        counts[nucleotide] += 1
    return counts

# Step 5: Compute the GC-content:
# Compute percentage as gc_percent => (gc_count / len(seq)) * 100.

def calculate_gc_content(sequence):
    """Calculate the GC content percentage of the sequence"""
    gc_count = sequence.count('G') + sequence.count('C')
    total_length = len(sequence)
    return (gc_count / total_length) * 100

# Step 6: Analyze DNA seq
def process_dna_seq():
    """Process a DNA sequence entered by the user"""
    
    # Get DNA sequence from user
    
    dna_sequence = input("Enter DNA sequence: ")

    
    # Clean and validate the sequence
    dna_sequence = dna_sequence.upper().replace(" ", "").replace("\n", "")
    
    if not is_valid_dna(dna_sequence):
        print("Invalid DNA sequence. Please check your input.")
        return
    
    # Process the sequence
    counts = count_nucleotides(dna_sequence)
    gc_content = calculate_gc_content(dna_sequence)
    
    # Display results
    print("\n" + "=" * 60)
    print("ANALYSIS RESULTS")
    print("=" * 60)
    
    print("\nNucleotide Counts:")
    print("-" * 40)
    for nucleotide, count in counts.items():
        print(f"  Counts for '{nucleotide}' = {count}")
    
    print("\nGC Content Analysis:")
    print("-" * 40)
    print(f"  GC content = {gc_content:.2f}%")
    print("=" * 60)


# To Execute the program
if __name__ == "__main__":
    # to test with the provided sequence
    test_sequence = """TGGCGCCCCCGACAGCCATGCGTACGGCAGGCCACAAGCCGTGGCGCGACCGACCACAGCGTGTCGGCGCTTTACGCGCGGCCGGACGGGACCTGGCAGGGCGAGGTTGTGTGGGCTGAGCCGCCGG
CTGCCCGGGAGCGGGCGGCAGCGGGGGGAACGGCGGGGATGAGCGGCCGCGCCGCCCGGCCCCCCCGGTACGCAGCAAGCCCGCCCTCGTCACGCGGGGTCCGGGGCAGGGTGGGGCAGGCGCTCCC
GTCGCCCGGCACGCTTGTGTCGGCGCTCGTTGGCGCCGGACCAGGCGGGTCAGCGGAAGGGCTGACCCGTGTGCCCGCCGGTTCCCGCAAGTGGCGGGCATCGCCCTGTCCGCCTGGGTCAGGGCAC
ACTATGCGGGGGAGTAGCTCGGAGTCTACCATTCGTTCGAGCGTCGCGCAGGAACATCGTCGAGCGCCCCCTCCCCTGAGCGCTCGCGGCCTTGGGCCCACGCCAGGGGGATCCCTCCTACGGCCGC
CCCGCCTGGGTAGCAACGCTGCGGCAACCGGGCGCCCGGAGCGCGAGTCACGTAGTGGGTCTACTTGGCCCGGGCCTGGATGGCGGGGCAGCCCCTCGCCCACCCGCCTCCCGCCCCGCCCCCCTGC
GTTACGCGGGGCCGGAGCGCTCGGCGGCGGGGCCGCGCACGGCCACCGCGCAGCGCGGCGGTcCGAGCGATCCGGCCGTCCCACGGCTGCCGGTGTCAGTCCAGTGCGTTCCGGGCTGGGGGACGAA"""
    
    # Clean the test sequence
    test_sequence = test_sequence.upper().replace(" ", "").replace("\n", "")
    
    print("Testing with provided sequence...")
    print(f"Sequence length: {len(test_sequence)}")
    
    # To Validate and process
    if is_valid_dna(test_sequence):
        counts = count_nucleotides(test_sequence)
        gc_content = calculate_gc_content(test_sequence)
        
        print("\nNucleotide Counts:")
        for nucleotide, count in counts.items():
            print(f"  Counts for '{nucleotide}' = {count}")
        
        print(f"\nGC content = {gc_content:.2f}%")
    
    # Uncheck below to run the interactive cell, and
    # process_dna_sequence()
```

```python
#  Task 1.2: DNA Sequence Analysis with Advanced Data Structures
# Purpose: Frequency count/analysis, reverse complement, and 
           # subsequence finding

# Step 1: Count Frequency
def frequency_count(dna_sequence):
    """
    Count frequency of each nucleotide using dict.
    
    Args:
        dna_sequence (str): DNA sequence to analyze
        
    Returns:
        dict: Frequency count of each nucleotide
    """
    
# Step 2: Initialize empty dictionary for efficient counting

    nucleotide_frequency = {}

    # count each nucleotide
    for nucleotide in dna_sequence:
        if nucleotide in nucleotide_frequency:
            nucleotide_frequency[nucleotide] += 1
        else:
            nucleotide_frequency[nucleotide] = 1
    
    return nucleotide_frequency

# Step 3: Next Step is to Reverse Complement: which means that A becomes T, 
# T becomes A, C becomes G, and G becomes C
def reverse_complement(dna_sequence):
    """
    Create reverse complement of DNA sequence.
    A pairs with T, C pairs with G.
    
    Args:
        dna_sequence (str): Original DNA sequence

    Returns:
        str: Reverse complement sequence
    """
    # Define complement mapping (or base-pair)
    base_pair = {
        'A': 'T',
        'T': 'A',
        'C': 'G',
        'G': 'C'
    }
            
# Step 4: For efficient computation/enumeration, Use list comprehension
# First reverse the sequence, then get complement of each base or nucleotide
    reverse_comp = ''.join([base_pair[base] for base in reversed(dna_sequence)])
    
    return reverse_comp


def find_subsequence(dna_sequence, target_subseq):
    """
    Find all starting positions of a subsequence in DNA.
    
    Args:
        dna_sequence (str): DNA sequence to search within
        target_subseq (str): Subsequence to look for
        
    Returns:
        list: List of starting indices where subsequence is found
    """
    found_positions = []
    target_length = len(target_subseq)
    
    # Search through sequence
    for i in range(len(dna_sequence) - target_length + 1):
        # Check if subsequence matches at found_position i
        if dna_sequence[i:i + target_length] == target_subseq:
            found_positions.append(i)
    return found_positions

# in the next line of cell, run_dna_sequence => analyze_dna_sequence
def run_dna_sequence(dna_sequence):
    """
    Main analysis functions for Task 1.2.
    
    Args:
        dna_sequence (str): DNA sequence to be analyzed
    """
    print("=" * 70)
    print("TASK 1.2: DNA SEQUENCE ANALYSIS")
    print("=" * 70)
    
#Sub-steps to finalizing full sequence analysis
    # Clean and prepare dna sequence
    cleaned_sequence = dna_sequence.upper().replace(" ", "").replace("\n", "")
    sequence_length = len(cleaned_sequence)

# would be to preview the first 50 bases 
    print(f"\nSequence Length: {len(cleaned_sequence)}")
    print(f"First 50 bases: {cleaned_sequence[:50]}...")
    
    # 1. Frequency Count
    print("\n" + "-" * 70)
    print("1. NUCLEOTIDE FREQUENCY COUNT")
    print("-" * 70)

    frequencies = frequency_count(cleaned_sequence)
    
    for nucleotide in ['A', 'T', 'C', 'G']:
        if nucleotide in frequencies:
            percentage = (frequencies[nucleotide] / len(cleaned_sequence)) * 100
            print(f"  {nucleotide}: {frequencies[nucleotide]:4d} occurrences ({percentage:5.2f}%)")
    
# 2. Reverse Complement
    print("\n" + "-" * 70)
    print("2. REVERSE COMPLEMENT")
    print("-" * 70)
    
    reverse_comp = reverse_complement(cleaned_sequence)

    print(f"\nOriginal sequence (first 50): {cleaned_sequence[:50]}")
    print(f"Reverse complement (first 50): {reverse_comp[:50]}")
    print(f"Reverse Complement Length: {len(reverse_comp)}")

# 3. Is to find Subsequence 'GAT'
    print("\n" + "-" * 70)
    print("3. SEARCH FOR SUBSEQUENCE 'GAT'")
    print("-" * 70)
    
    target_seq = 'GAT'
    positions = find_subsequence(cleaned_sequence, target_seq)
    
    print(f"\nSubsequence '{target_seq}' found at {len(positions)} positions:")

    if len(positions) > 0:
        print(f"First 10 Positions: {positions[:10]}")
        if len(positions) > 10:
            print(f"... and {len(positions) - 10} more positions")
    else:
        print("  No Subsequence found")
    
    print("\n" + "=" * 70)

# Final Step:
# Next is to ; Test with provided test_sequence (Sample)
if __name__ == "__main__":
    test_dna_sequence = """ATCGATCGAACATCGATCATCATCGATCGAACATCGATCATCGATCGAAATCGATCGAACCGAACGATCGATCGAACATATCGATATCGATCGAACCGATCGAACATCGAACCATCGATCGAAATCG
ATCGAACCGAACGATCGATCGAACATATCGATATCGATCGAACCGATCGATCGAACATCGATCATCGATCGAAATCGATCGAACCGAACGATCGATCGAACATATCGATATCGATCGAACCGATCGA
ACATCGAACCATCATCGAACATCGAAATCGATCGAACATCGATCATCGATCGAAATCGATCGAACCGAACGATCGATCGAACATATCGATATCGATCGAACCGATCGAACATCGAACCATCCCATCA
TCGATCGAACATCGATCATCGATCGAAA"""
    run_dna_sequence(test_dna_sequence)
```

```python
# Task 1.3: DNA Sequence Mutation Analysis
# Purpose: To Validate sequence, find nucleotide runs, identify mutation positions

# Step 1: to validate & clean dna sequence
def validate_and_clean_dna_sequence(sequence):
    """
    Keep Validated DNA sequence and remove invalid characters found.
    
    Args:
        sequence (str): DNA sequence to validate, possibly with errors
        
    Returns:
        tuple: (is_valid, cleaned_sequence, invalid_found)
    """
    valid_nucleotides = {'A', 'T', 'C', 'G'}
    invalid_found = set()
    cleaned_sequence = ""
    
# Step 2:  Check each character
    for char in sequence.upper():
        if char in valid_nucleotides:
            cleaned_sequence += char
        elif char not in {' ', '\n', '\t'}:  
            invalid_found.add(char)
    
    is_valid = len(invalid_found) == 0
    
    return is_valid, cleaned_sequence, invalid_found

#Step 3: Is to Count consecutive nucleotide Run
def count_longest_nucleotide_run(sequence, targeted_nucleotide):
    """
    Count the longest consecutive run of a specific nucleotide in the sequence.
    
    Args:
        sequence (str): DNA sequence
        targeted_nucleotide (str): Nucleotide to look for ('T')
        
    Returns:
        tuple: (max_run_length, start_position)
    """
    max_run = 0       # Length of the longest sequence run so far
    current_run = 0     # Length of the current sequence being counted
    max_run_start = -1     # Starting index of the longest dna sequence found
    current_run_start = -1     # Starting index of the sequence currently being run
    
    i = 0
    while i < len(sequence):
        if sequence[i] == targeted_nucleotide:
            if current_run == 0:
                current_run_start = i
            current_run += 1
            
# Update maximum, if current run is longer (.i.e. using a control flow statement approach)
            if current_run > max_run:
                max_run = current_run
                max_run_start = current_run_start
        else:
            current_run = 0
        
        i += 1
    
    return max_run, max_run_start

# Step 4: Find Mutations .i.e. to Identify all the positions in the DNA sequence
def find_mutation_positions(sequence, targeted_nucleotide):
    """
    Find all positions where mutation targeting nucleotide is possible.
    (All positions that the nucleotide is NOT already the target)
    
    Args:
        sequence (str): DNA sequence
        targeted_nucleotide (str): Target nucleotide for mutation
        
    Returns:
        list: Positions where mutation is possible
    """
    mutation_positions = []
    
    for i in range(len(sequence)):
        if sequence[i] != targeted_nucleotide:
            mutation_positions.append(i)
    
    return mutation_positions

# for the next line of cell, analyze dna mutation => evaluate_dna_mutation
def evaluate_dna_mutations(sequence):
    """
    Main basis for Task 1.3 - Complete mutation analysis.
    
    Args:
        sequence (str): DNA sequence to analyze
    """
    print("=" * 70)
    print("TASK 1.3: DNA SEQUENCE MUTATION ANALYSIS")
    print("=" * 70)

def evaluate_dna_mutations(sequence, target_nucleotide='C'):  
    target_nucleotide = 'C'
    
        # added target_nucleotide with given base 'C'

# Break down of the steps and analysis
# Step 1 break down: Validate and clean Sequence
    print("\n" + "-" * 70)
    print("STEP 1: SEQUENCE VALIDATION")
    print("-" * 70)
    
    is_valid, cleaned_seq, invalid_found = validate_and_clean_dna_sequence(sequence)
    
    if not is_valid:
        print(f"\nInvalid characters found: {invalid_found}")
        print(f"Original length: {len(sequence)}")
        print(f"Cleaned Sequence length: {len(cleaned_seq)}")
        print("\nSequence has been cleaned and will be re-analyzed")
    else:
        print("\nSequence is valid - no invalid characters found")
        cleaned_seq = sequence.upper().replace(" ", "").replace("\n", "")
    
# Step 2: is to sort out & Re-analyze cleaned sequence (SUMMARY)
    print("\n" + "-" * 70)
    print("STEP 2: CLEANED SEQUENCE ANALYSIS")
    print("-" * 70)
    print(f"\nSequence length: {len(cleaned_seq)}")
    print(f"First 60 bases: {cleaned_seq[:60]}")
    
# Step 3: Count Longest given nucleotide, 'T' Run
    print("\n" + "-" * 70)
    print("STEP 3: LONGEST CONSECUTIVE 'T' RUN")
    print("-" * 70)
    
    target_for_run = 'T'  #Nucleotide targeted for run
    max_run, start_pos = count_longest_nucleotide_run(cleaned_seq, target_for_run)
    
    print(f"\nLongest consecutive run of '{target_for_run}':")
    print(f"Length: {max_run} nucleotides")
    print(f"Starting position: {start_pos}")   #start_pos = start_position
    
    if start_pos >= 0:
        end_pos = start_pos + max_run   #end_pos = end_position
        context_start = max(0, start_pos - 10)
        context_end = min(len(cleaned_seq), end_pos + 10)
        
        print(f"\nSequence Context:")
        print(f"  ...{cleaned_seq[context_start:start_pos]}"
              f"[{cleaned_seq[start_pos:end_pos]}]"
              f"{cleaned_seq[end_pos:context_end]}...")
    
# Step 4: Look for Mutation Positions and possibilities for 'C'
    print("\n" + "-" * 70)
    print("STEP 4: POSSIBLE MUTATION POSITIONS FOR 'C'")
    print("-" * 70)


    
    target_nucleotide = 'C'
    mutation_pos = find_mutation_positions(cleaned_seq, target_nucleotide)   #mutation_pos -> mutation_position


    print(f"\nTotal positions where mutation to '{target_nucleotide}' is possible: {len(mutation_pos)}")
    print(f"\nFirst 20 Mutation Positions: {mutation_pos[:20]}")
    
    if len(mutation_pos) > 20:
        print(f"... and {len(mutation_pos) - 20} more positions")
    
# Step 5: Mutational statistics
    print("\n" + "-" * 70)
    print("MUTATIONAL STATISTICS")
    print("-" * 70)
    
    current_c_count = cleaned_seq.count(target_nucleotide)
    potential_c_count = len(mutation_pos) + current_c_count
    
    print(f"\nCurrent '{target_nucleotide}' Count: {current_c_count}")
    print(f"Maximum Possible '{target_nucleotide}' after mutations: {potential_c_count}")
    print(f"Mutation Possibility: {(len(mutation_pos)/len(cleaned_seq))*100:.2f}% of sequence")
    
    print("\n" + "=" * 70)


# Test with provided sequence
if __name__ == "__main__":
    test_sequence = """ACTAACATCCAGTTAATAATTTATACATAGTTTAAGAATATACTACATTAAAAATACGAACTAAGACTGTATTACGCTCCCATCTAATCTATATTTTGTGTTCTATAAGATCCTTTATTAATATAG
TATACACTGTAATTATTACAACCAATCTGTGTAAGTAGGTTAATGCGATATTTGTTTGTTCTAAATCAATGATAACGTATAATTTCTATTATTCTAAATTCTCTCCTATCTGTCTATTTCAGCAGG
ATAAAGATAGAGGTATTCTTCCAATGTTTAATAATATGATAGTATCAATTATAAGTGGAATTTGTAATTAATTACCTATCAATTTCTACGATGATATGTCAGGACCCCTCGTAAACCGCTTCGTAA
TTTTAAAACGTACCATTGTTAATATCCGTGAGCGTTATGAAACTAGGTCGCTACCTGAATTATTGTcAcATGTTTTGCGTGAAAATCGTAGTTCGAATTGTCTTTAATAAGTCTTGTCAATTCGAA
CGTATACTCTTAAACTAAAAGGAATGAATTAAGAAAGTACAATTCAATACGTACGAAATAATAAAACCCGATCGCGTGATGCGTTCCCGAAATTTGGTAACAGTGACAACATACGACAACATATTT
GTGGGTATTAATCCTTTATAAGTCTTCTATATATGATTAGAAATAACAGATGTAACCAATAATTGCTAAAATAAAGATAGAGGTATTCTTCCAATGTTTAATAATATGATAGTATCAATTATAAGTGGAATTTGTAATTAATTACCTATCAATTTCTACGATGATATGTCAGGACCCCTCGTAAACCGCTTCGTAA
TTTTAAAACGTACCATTGTTAATATCCGTGAGCGTTATGAAACTAGGTCGCTACCTGAATTATTGTcAcATGTTTTGCGTGAAAATCGTAGTTCGAATTGTCTTTAATAAGTCTTGTCAATTCGAA
CGTATACTCTTAAACTAAAAGGAATGAATTAAGAAAGTACAATTCAATACGTACGAAATAATAAAACCCGATCGCGTGATGCGTTCCCGAAATTTGGTAACAGTGACAACATACGACAACATATTT"""
    
    evaluate_dna_mutations(test_sequence)
```

```python
# Task 1.4: Debugging Exercise - dna sequence analysis
# Purpose: Fix bugs in nucleotide frequency and triplet analysis

# Bug Analysis and corrected codes

"""
BUG ANALYSIS:

Bug #1 in nucleotide_frequency():
>> PROBLEM: Dictionary only has A, T, C, G but sequence contains spaces
>> ERROR: KeyError when encountering space character
>> SOLUTION: Gauze out non-nucleotide characters before counting

Bug #2 in most_common_triplet() -> First version:
>> PROBLEM: Uses range(len(dna)) which goes too far
>> ERROR: Creates invalid triplets at sequence end
>> SOLUTION: Use range(len(dna) -> 2) to ensure valid 3-char triplets

Bug #3 in most_common_triplet() -> Second version:
>> PROBLEM: Uses step (3) (range(0, len(dna)-2, 3))
>> ERROR: Misses overlapping triplets (only scan non-overlapping)
>> SOLUTION: Use step (1) to capture ALL possible triplets

Bug #4: Function defined twice
>> PROBLEM: most_common_triplet defined twice with different logic
>> ERROR: Second definition overwrites first
>> SOLUTION: Keep only one corrected version
"""


def nucleotide_frequency(dna):
    """
    Calculate frequency of each nucleotide (A, T, C, G) in DNA sequence.
    Ignores non-nucleotide characters properly.
    
    Args:
        dna (str): DNA sequence
        
    Returns:
        dict: Frequency count of each nucleotide
    """
# Init. frequency dictionary
    freq = {'A': 0, 'T': 0, 'C': 0, 'G': 0}
    
# Count only valid nucleotides (ignore spaces and other characters)
    for char in dna:
        if char in freq:  # .i.e. Only count if it's a valid nucleotide
            freq[char] += 1
    
    return freq


def most_common_triplet(dna):
    """
    Find the most common triplet (3-nucleotide sequence) in DNA.
    FIXED: Only captures ALL possible overlapping triplets.
    
    Args:
        dna (str): DNA sequence
        
    Returns:
        tuple: (most common triplet, triplet count)
    """
# Clean sequence first: thereby removing non-nucleotide characters (just keeping valid nucleotide)
    valid_nucleotides = {'A', 'T', 'C', 'G'}
    cleaned_dna = ''.join([char for char in dna if char in valid_nucleotides])
    
# Init. triplet counter  (init. => initialize)
    triplet_count = {}
    
# Count/capture all overlapping triplets using a sliding window (step = 1, not 3)
    for i in range(len(cleaned_dna) - 2):  # Ensuring it doesn't go past end
        triplet = cleaned_dna[i:i + 3]
        
# Only count if there is a full 3-character triplet
        if len(triplet) == 3:
            if triplet in triplet_count:
                triplet_count[triplet] += 1
            else:
                triplet_count[triplet] = 1
    
# Find triplet with maximum count .i.e. (looking for the triplet with the highest count)
    if triplet_count:
        max_triplet = max(triplet_count, key=triplet_count.get)
        return max_triplet, triplet_count[max_triplet]
    else:
        return None, 0


def analyze_dna_debug():
    """
    Main function to test the debugged code.
    """
    print("=" * 70)
    print("TASK 1.4: DEBUGGING -> CORRECTED VERSION")
    print("=" * 70)
    
# Test sequence from task: Original (buggy) sequence used for testing
    dna_sequence = """AAAATACAATGTACATTCTTCGTATTATAGACATTTAGCTTAAAAATATACCAAAGATTTAATTTTATATTAATACATCAATATTTAATCTGCTGTATATGCGTTTATTTATAGGGGATGGAAATAT
ACTTAGTCATTATAAAGAAAATTTTGCTTGATATAATTATGACATATAAGTCTGTACAAATCACTTGATTTATTTGACACAATGCCAGCTCAAGTAATGACTATTTGGTACATCTTAAGAAAGAGTA
AGCCGGACTTTGAGTTTGACAGATTTATTTAAATACATTCATATATCATTCGATTTATTTTTCTTATTTACCTTTTTTAGGACCTTATTTTTTATAATTAAAATCTTTAATTTAAAATTTAAAATCTTTCAAAGTTACAAATTTAAGGACA
ACTTTATACTATAAAGTATACCTTTTTTAATTATTTAATAAGTTATACAAGAATTATACCATTAACTCACAAACTTTAGCTTATAAATACTTGATTAATTGACTAGATTCATTTT"""
    
# Clean dna sequence
    cleaned_seq = dna_sequence.replace(" ", "").replace("\n", "")
    
    print(f"\nSequence length: {len(cleaned_seq)}")
    
# Test nucleotide frequency
    print("\n" + "-" * 70)
    print("NUCLEOTIDE FREQUENCY ANALYSIS")
    print("-" * 70)
    
    frequencies = nucleotide_frequency(cleaned_seq)
    
    total_freq = sum(frequencies.values())     #total_freq => total frequencies
    print(f"\nTotal nucleotides counted: {total_freq}")
    print(f"Original sequence length: {len(cleaned_seq)}")
    
    for nucleotide in ['A', 'T', 'C', 'G']:
        count = frequencies[nucleotide]
        percentage = (count / total_freq) * 100 if total_freq > 0 else 0
        print(f"  {nucleotide}: {count:4d} ({percentage:5.2f}%)")
    
# Test most common triplet (Triplet analysis)
    print("\n" + "-" * 70)
    print("MOST COMMON TRIPLET ANALYSIS")
    print("-" * 70)
    
    triplet, count = most_common_triplet(cleaned_seq)
    
    if triplet:
        print(f"\nMost common triplet: {triplet}")
        print(f"Occurs: {count}")
        
# Calculate percentage
        total_possible_triplets = len(cleaned_seq) - 2
        percentage = (count / total_possible_triplets) * 100
        print(f"Frequency: {percentage:.2f}% of all possible triplets")
    else:
        print("\nNo triplets found (sequence too short)")
    
# Additional test with custom sequence
    print("\n" + "=" * 70)
    print("ADDITIONAL TEST - Custom DNA Sequence")
    print("=" * 70)
    
    test_seq = "ATGATGATGATGCCCCCCCCCAAAAAAA"   # test_seq = test sequence (sample)
    print(f"\nTest sequence: {test_seq}")
    print(f"Length: {len(test_seq)}")
    
    freq_test_seq = nucleotide_frequency(test_seq)
    triplet_test_seq, count_test_seq = most_common_triplet(test_seq)
    
    print("\nNucleotide frequencies:")
    for nuc, cnt in freq_test_seq.items():      # nuc = nucleotide, cnt = count
        print(f"  {nuc}: {cnt}")
    
    print(f"\nMost common triplet: {triplet_test_seq} (count: {count_test_seq})")
    
    print("\n" + "=" * 70)
    print("BUGS FIXED:")
    print("-" * 70)
    print("1. Added character validation in nucleotide_frequency()")
    print("2. Fixed range in triplet counting (len-2 instead of len)")
    print("3. Changed step from 3 to 1 to capture All overlapping triplets")
    print("4. Removed duplicate function definition")
    print("5. Added sequence cleaning before triplet analysis")
    print("=" * 70)


# Run the debugged version
if __name__ == "__main__":
    analyze_dna_debug()
```

```python
# Task 2.1: DNA Analysis and Phylogenetic Tree Construction
# Purpose: Analyze DNA → RNA → Protein, & Complete bioinformatics pipeline without external MSA tools

# BIOINFORMATICS PIPELINE & DATA USAGE

# Biopython utilities:

# Sequence Handling and Records
from Bio import SeqIO
from Bio.Seq import Seq
from Bio.SeqRecord import SeqRecord

# Sequence Alignment Tools
from Bio import Align
from Bio.Align import MultipleSeqAlignment

# Phylogenetic Tree Construction
from Bio.Phylo.TreeConstruction import DistanceCalculator, DistanceTreeConstructor

# Data Processing and System Utilities
from collections import Counter
import warnings

warnings.filterwarnings('ignore')
```

```python

def transcribe_dna_to_rna(dna_sequence):
    """
    Transcribe DNA sequence to RNA by replacing T with U.
    
    Args:
        dna_sequence (Seq): DNA sequence object
        
    Returns:
        Seq: RNA sequence object
    """
# Using built-in transcription method of Biopython
    rna_sequence = dna_sequence.transcribe()
    return rna_sequence


def translate_rna_to_protein(rna_sequence):
    """
    Translate RNA sequence to protein using standard genetic code.
    
    Args:
        rna_sequence (Seq): RNA sequence object
        
    Returns:
        Seq: Protein sequence object
    """
# Use Biopython's built-in translation method
    protein_sequence = rna_sequence.translate()
    return protein_sequence


def calculate_amino_acid_frequency(protein_sequence):
    """
    Calculate frequency of each amino acid in protein sequence.
    
    Args:
        protein_sequence (Seq): Protein sequence object
        
    Returns:
        dict: Dictionary with amino acid frequencies
    """
# Convert to string and count using 'Counter'
    amino_acid_counts = Counter(str(protein_sequence))
    

# Sort by amino acid letter for consistent display
    sorted_counts = dict(sorted(amino_acid_counts.items()))
    
    return sorted_counts

# Align two sequences globally, return the best match and score
def perform_pairwise_alignment(seq1, seq2):
    """
    Perform global pairwise alignment between two sequences.
    
    Args:
        seq1 (Seq): First sequence
        seq2 (Seq): Second sequence
        
    Returns:
        tuple: (alignment, score) 
    """
# Create Pairwise-aligner object
    aligner = Align.PairwiseAligner()
    
# Set alignment limit for global alignment
    aligner.mode = 'global'
    aligner.match_score = 2
    aligner.mismatch_score = -1
    aligner.open_gap_score = -2
    aligner.extend_gap_score = -0.5
    
# Perform alignment
    alignments = aligner.align(str(seq1), str(seq2))
    
# Get best alignment
    best_alignment = alignments[0]
    score = best_alignment.score
    
    return best_alignment, score

# Generate distance matrix from list of sequence record
def create_distance_matrix(sequences):
    """
    Generate distance matrix from DNA sequences.
    
    Args:
        sequences (list): List of SeqRecord objects
        
    Returns:
        DistanceMatrix: Distance matrix object
    """
# Ensure all sequences are same length for alignment
    max_length = max(len(seq.seq) for seq in sequences)
    
# Pad sequences to same length if needed
    padded_seqs = []
    for seq_record in sequences:
        seq_str = str(seq_record.seq)
# Pad with gaps (if needed)
        if len(seq_str) < max_length:
            seq_str = seq_str + '-' * (max_length - len(seq_str))
        
        padded_seq = SeqRecord(Seq(seq_str), id=seq_record.id)
        padded_seqs.append(padded_seq)
    
# Create MultipleSeqAlignment object
    alignment = MultipleSeqAlignment(padded_seqs)
    
# Calculate distance matrix using identity model
    calculator = DistanceCalculator('identity')
    distance_matrix = calculator.get_distance(alignment)
    
    return distance_matrix

# Build a Neighbor-Joining phylogenetic tree (using DistanceTreeConstructor).
def construct_phylogenetic_tree(distance_matrix):
    """
    Build Neighbor-Joining phylogenetic tree from distance matrix.
    
    Args:
        distance_matrix: Distance matrix object
        
    Returns:
        Tree: Phylogenetic tree object
    """
# Create tree constructor
    constructor = DistanceTreeConstructor()
    
# Build tree using Neighbor-Joining method
    tree = constructor.nj(distance_matrix)
    
    return tree

fasta_file = 'Downloads/Python_Final_assessment/sequences_2.1.fasta'

def analyze_sequences_complete(fasta_file):
    """
    Main function to perform complete DNA analysis pipeline.
    
    Args:
        fasta_file (str): Path to FASTA file
    """
    print("=" * 80)
    print("TASK 2.1: DNA ANALYSIS AND PHYLOGENETIC TREE CONSTRUCTION")
    print("=" * 80)
    
# Load sequences from FASTA file
    sequences = list(SeqIO.parse(fasta_file, "fasta"))
    
    if len(sequences) == 0:
        print("\nError: No sequences found in FASTA file")
        return
    
    print(f"\nLoaded {len(sequences)} sequences from {fasta_file}")
    
# Store all sequences for later use
    all_sequences = []
    
    # STEP 1-3: Analyze each sequence
    print("\n" + "=" * 80)
    print("STEP 1-3: DNA, RNA, PROTEIN ANALYSIS")
    print("=" * 80)
    
    for idx, seq_record in enumerate(sequences, 1):
        print(f"\n{'-' * 80}")
        print(f"SEQUENCE {idx}: {seq_record.id}")
        print(f"{'-' * 80}")
        
# Get DNA sequence
        dna_seq = seq_record.seq
        print(f"\nDNA Length: {len(dna_seq)} bp")   # bp = base pair
        print(f"DNA (first 60 bp): {str(dna_seq)[:60]}...")
        
 # Transcribe to RNA
        rna_seq = transcribe_dna_to_rna(dna_seq)
        print(f"\nRNA Length: {len(rna_seq)} nt")
        print(f"RNA (first 60 nt): {str(rna_seq)[:60]}...")
        
 # Translate to Protein
        protein_seq = translate_rna_to_protein(rna_seq)
        print(f"\nProtein Length: {len(protein_seq)} aa")
        print(f"Protein (first 60 aa): {str(protein_seq)[:60]}...")
        
# Calculate amino acid frequencies
        aa_frequencies = calculate_amino_acid_frequency(protein_seq)
        
        print(f"\nAmino Acid Frequencies:")
        print(f"{'Amino Acid':<15} {'Count':<10} {'Percentage':<10}")
        print("-" * 40)
        
        total_aa = len(protein_seq)
        for aa, count in aa_frequencies.items():
            if aa != '*':  # Skip stop codon
                percentage = (count / total_aa) * 100
                print(f"{aa:<15} {count:<10} {percentage:>6.2f}%")
        
# Store sequence for phylogenetic analysis
        all_sequences.append(seq_record)

        
    
# STEP 4: Pairwise Alignment
    if len(sequences) >= 2:
        print("\n" + "=" * 80)
        print("STEP 4: PAIRWISE ALIGNMENT")
        print("=" * 80)
        
        seq1 = sequences[0]
        seq2 = sequences[1]
        
        print(f"\nAligning: {seq1.id} vs {seq2.id}")
        
        alignment, score = perform_pairwise_alignment(seq1.seq, seq2.seq)
        
        print(f"\nAlignment Score: {score:.2f}")
        print("\nAlignment (first 200 characters):")
        print("-" * 80)
        
# Display alignment in a formatted way
        alignment_str = str(alignment)
        lines = alignment_str.split('\n')
        for i, line in enumerate(lines[:10]):  # Show first 10 lines
            print(line)
        
        if len(lines) > 10:
            print(f"\n... ({len(lines) - 10} more lines)")
    
# STEP 5: Phylogenetic Tree Construction
    if len(sequences) >= 2:
        print("\n" + "=" * 80)
        print("STEP 5: PHYLOGENETIC TREE CONSTRUCTION")
        print("=" * 80)
        
        print("\nConstructing phylogenetic tree using Neighbor-Joining method...")
        
        try:
# Create distance matrix
            distance_matrix = create_distance_matrix(all_sequences)
            
            print("\nDistance Matrix:")
            print("-" * 80)
            print(distance_matrix)
            
            # Construct phylogenetic tree
            tree = construct_phylogenetic_tree(distance_matrix)
            
            print("\nPhylogenetic Tree (ASCII format):")
            print("-" * 80)
            
# Display tree
            from Bio import Phylo
            import io
            
 # Capture tree output
            output = io.StringIO()
            Phylo.draw_ascii(tree, file=output)
            tree_str = output.getvalue()
            print(tree_str)
            
            print("\nTree interpretation:")
            print("- Branch lengths represent evolutionary distance")
            print("- Closer branches indicate more similar sequences")
            print("- Tree shows inferred evolutionary relationships")
            
        except Exception as e:
            print(f"\nError constructing tree: {e}")
            print("Note: Sequences may need to be of similar lengths for tree construction")
    
    print("\n" + "=" * 80)
    print("ANALYSIS COMPLETE")
    print("=" * 80)
    

# Primary execution
if __name__ == "__main__":
    
    fasta_file = 'Downloads/Python_Final_assessment/sequences_2.1.fasta'
    print("Starting DNA Analysis Pipeline...")
    print(f"Looking for file: {fasta_file}\n")

# Run analysis
    analyze_sequences_complete(fasta_file)
   
```

```python
# Task 2.2: Cancer-Associated Skeletal Muscle Wasting Analysis
# Purpose: Machine learning prediction from metabolite data

import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import os
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import classification_report, roc_curve, auc, accuracy_score
import warnings
warnings.filterwarnings('ignore')
```

```python
filename = 'Downloads/Python_Final_assessment/metabolite 2.2.xlsx'

# Set plot style
sns.set_style("whitegrid")
plt.rcParams['figure.dpi'] = 100

# Data Description (Loading and data preparation)

def load_and_prepare_data(filename):
    """
    Load and prepare metabolite data from Excel file.
    Handles data that needs to be transposed.
    
    Args:
        filename (str): Path to Excel file
        
    Returns:
        DataFrame: Loaded and cleaned data
    """
    print("=" * 80)
    print("PART 1: DATA PREPARATION")
    print("=" * 80)
    
# Check if file exists
    if not os.path.exists(filename):
        print(f"\n ERROR: File not found at '{filename}'")
        return None
    
    print(f"\n File found: {filename}")
    
# Load Excel file
    try:
        if filename.endswith('.xlsx'):
            df_raw = pd.read_excel(filename, engine='openpyxl')
        else:
            df_raw = pd.read_excel(filename)
        
        print(f" Successfully loaded data from Excel file")
        print(f"  Raw data shape: {df_raw.shape}")
        
    except Exception as e:
        print(f"\n ERROR: Could not read Excel file: {e}")
        return None
    
    
# Check first few column names
    print(f"\nFirst column name: {df_raw.columns[0]}")
    print(f"First few values in first column:\n{df_raw.iloc[:5, 0]}")

# Determine if there is a need to transpose
# since column headers look like sample IDs (contain PIF, NET, etc.), then transpose
    sample_indicators = ['PIF', 'NET', 'NETL', 'NETCR']
    needs_transpose = any(indicator in str(df_raw.columns[1]) for indicator in sample_indicators)
    
    if needs_transpose:
        print("\n Data structure detected: Metabolites as rows, samples as columns")
        print("   Transposing data so samples are rows...")
        
# Set first column as index (metabolite names)
        df_raw = df_raw.set_index(df_raw.columns[0])
        
# Transpose: now samples are rows, metabolites are columns
        df = df_raw.T
        
# Reset index to make sample IDs a regular column
        df = df.reset_index()
        df = df.rename(columns={'index': 'Sample_ID'})
        
        print(f" Transposed data shape: {df.shape}")
        
    else:
        print("\n Data structure: Samples as rows (no transpose needed)")
        df = df_raw.copy()
    
# Display basic information
    print("\n" + "-" * 80)
    print("PROCESSED DATA STRUCTURE")
    print("-" * 80)
    print(f"\nDataset shape: {df.shape}")
    print(f"Number of samples: {len(df)}")
    print(f"Number of columns: {len(df.columns)}")
    
    print("\nFirst 5 column names:")
    for i, col in enumerate(df.columns[:5], 1):
        print(f"  {i}. {col}")
    
# Identify numeric columns (metabolites concentrations)
    numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()
    non_numeric_cols = [col for col in df.columns if col not in numeric_cols]
    
    print(f"\nNumeric columns (metabolites): {len(numeric_cols)}")
    print(f"Non-numeric columns: {non_numeric_cols}")
    
# Check for missing values in numeric columns
    print("\n" + "-" * 80)
    print("MISSING VALUES CHECK")
    print("-" * 80)
    
    missing = df[numeric_cols].isnull().sum()
    total_missing = missing.sum()
    
    if total_missing == 0:
        print("\n No missing values found in metabolite data")
    else:
        print(f"\n Total missing values: {total_missing}")
        print("\nMissing values per metabolite:")
        missing_cols = missing[missing > 0]
        for col, count in missing_cols.items():
            print(f"  {col}: {count} ({count/len(df)*100:.2f}%)")
        
# Handling missing values in X by filling with column means
        print("\nHandling missing values by filling with column mean...")
        df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].mean())
        print(" Missing values handled")
    
# Create or identify Class column
    # Look for class labels in the data or create based on sample names
    if 'Class' not in df.columns and 'class' not in df.columns:
        print("\n" + "-" * 80)
        print("CLASS LABEL DETECTION")
        print("-" * 80)
        
# Create class based on sample naming convention
# Assume: PIF samples = Cachexic (1), NET samples = Control (0)
# lambda = λ        
        if 'Sample_ID' in df.columns:
            df['Class'] = df['Sample_ID'].apply(
                lambda x: 1 if 'PIF' in str(x) else 0
            )
            print("\n Class labels created based on sample IDs:")
            print("  - PIF samples → Cachexic (Class = 1)")
            print("  - NET samples → Control (Class = 0)")
        else:
            print("\n [WARNING]: Cannot determine class labels")
            print("  Creating default classes based on first/second half of data")
            df['Class'] = [1 if i < len(df)//2 else 0 for i in range(len(df))]

# i Identifying target column and feature columns
# Ensure Class column exists
    class_col = 'Class' if 'Class' in df.columns else 'class'
    
# Descriptive statistics
    print("\n" + "-" * 80)
    print("DESCRIPTIVE STATISTICS (First 5 metabolites)")
    print("-" * 80)
    numeric_cols_display = [col for col in numeric_cols if col != class_col][:5]
    if numeric_cols_display:
        print(df[numeric_cols_display].describe())
    
# Class distribution
    print("\n" + "-" * 80)
    print("CLASS DISTRIBUTION")
    print("-" * 80)
    print(f"\nTarget variable: '{class_col}'")
    print(df[class_col].value_counts())
    print(f"\nCachexic (1): {(df[class_col] == 1).sum()}")
    print(f"Control (0): {(df[class_col] == 0).sum()}")
    
    return df

# Data Visualization

def visualize_data(df):
    """
    Create comprehensive data visualizations.
    
    Parameters:
        df (DataFrame): Metabolite data
    """
    print("\n" + "=" * 80)
    print("PART 2: DATA VISUALIZATION")
    print("=" * 80)
    
# Identify class and feature columns
    class_col = 'Class' if 'Class' in df.columns else 'class'
    
# Get numeric columns (excluding Sample_ID and Class)
    numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()
    feature_cols = [col for col in numeric_cols if col != class_col]
    
    print(f"\nUsing '{class_col}' as target variable")
    print(f"Number of features (metabolites): {len(feature_cols)}")
    print(f"Feature names (first 10): {feature_cols[:10]}")
    
# 1. Distribution of metabolites (histogram)
    print("\n1. Creating metabolite distribution histograms...")
    
    fig, axes = plt.subplots(3, 4, figsize=(15, 10))
    axes = axes.ravel()
    
    n_plots = min(12, len(feature_cols))
    
    for i in range(n_plots):
        axes[i].hist(df[feature_cols[i]].dropna(), bins=20, alpha=0.7, 
                    color='steelblue', edgecolor='black')
        axes[i].set_title(f'{feature_cols[i][:20]}', fontsize=9)
        axes[i].set_xlabel('Concentration', fontsize=8)
        axes[i].set_ylabel('Frequency', fontsize=8)
        axes[i].tick_params(labelsize=7)
    
# Hide unused subplots
    for i in range(n_plots, 12):
        axes[i].axis('off')
    
    plt.tight_layout()
    plt.savefig('metabolite_distributions.png', dpi=300, bbox_inches='tight')
    print(" Saved: metabolite_distributions.png")
    plt.show()
    
# 2. Boxplot comparison between classes
    print("\n2. Creating class comparison boxplots...")
    
    fig, axes = plt.subplots(2, 3, figsize=(15, 8))
    axes = axes.ravel()
    
    selected_metabolites = feature_cols[:6]
    
    for i, col in enumerate(selected_metabolites):
        cachexic_data = df[df[class_col] == 1][col].dropna()
        control_data = df[df[class_col] == 0][col].dropna()
        
        box_data = [cachexic_data, control_data]
        
        bp = axes[i].boxplot(box_data, labels=['Cachexic', 'Control'], 
                            patch_artist=True, widths=0.6)
        
        colors = ['lightcoral', 'lightblue']
        for patch, color in zip(bp['boxes'], colors):
            patch.set_facecolor(color)
        
        axes[i].set_title(f'{col[:25]}', fontsize=10)
        axes[i].set_ylabel('Concentration', fontsize=8)
        axes[i].grid(axis='y', alpha=0.3)
        axes[i].tick_params(labelsize=8)
    
    plt.tight_layout()
    plt.savefig('class_comparison_boxplots.png', dpi=300, bbox_inches='tight')
    print(" Saved: class_comparison_boxplots.png")
    plt.show()
    
# 3. Correlation heatmap
    print("\n3. Creating correlation heatmap...")
    
    plt.figure(figsize=(14, 12))
    
# Calculate correlation matrix
    corr_data = df[feature_cols].copy()
    corr_matrix = corr_data.corr()
    
# Create heatmap
    sns.heatmap(corr_matrix, 
                cmap='coolwarm', 
                center=0,
                square=True,
                linewidths=0.5,
                cbar_kws={"shrink": 0.8},
                annot=False)
    
    plt.title('Correlation Heatmap of Metabolites', fontsize=14, fontweight='bold')
    plt.xticks(fontsize=6, rotation=90)
    plt.yticks(fontsize=6)
    plt.tight_layout()
    plt.savefig('correlation_heatmap.png', dpi=300, bbox_inches='tight')
    print(" Saved: correlation_heatmap.png")
    plt.show()
    
# 4. PCA Analysis
    print("\n4. Performing PCA dimensionality reduction...")
    
# Prepare data for PCA
    X_pca = df[feature_cols].copy()
    
# Standardize features (Create a StandardScaler to scale features)
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X_pca)
    
# Perform PCA
    pca = PCA(n_components=2)
    pca_components = pca.fit_transform(X_scaled)
    
# Create PCA DataFrame
    pca_df = pd.DataFrame(
        data=pca_components,
        columns=['PC1', 'PC2']
    )
    pca_df['Class'] = df[class_col].values
    
# Plot PCA
    plt.figure(figsize=(10, 8))
    
    colors = ['red', 'blue']
    labels = ['Cachexic', 'Control']
    markers = ['o', 's']
    
    for class_value, color, label, marker in zip([1, 0], colors, labels, markers):
        indices = pca_df['Class'] == class_value
        plt.scatter(
            pca_df.loc[indices, 'PC1'],
            pca_df.loc[indices, 'PC2'],
            c=color,
            label=label,
            alpha=0.6,
            s=100,
            marker=marker,
            edgecolors='black',
            linewidth=0.5
        )
    
    plt.xlabel(f'PC1 ({pca.explained_variance_ratio_[0]:.2%} variance)', fontsize=12)
    plt.ylabel(f'PC2 ({pca.explained_variance_ratio_[1]:.2%} variance)', fontsize=12)
    plt.title('PCA: Metabolite Data Projection', fontsize=14, fontweight='bold')
    plt.legend(fontsize=11, loc='best')
    plt.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.savefig('pca_scatter_plot.png', dpi=300, bbox_inches='tight')
    print(" Saved: pca_scatter_plot.png")
    plt.show()
    
    print(f"\nPCA Results:")
    print(f" -> PC1 explains {pca.explained_variance_ratio_[0]:.2%} of variance")
    print(f" -> PC2 explains {pca.explained_variance_ratio_[1]:.2%} of variance")
    print(f" -> Total variance explained: {sum(pca.explained_variance_ratio_):.2%}")


def build_machine_learning_model(df):
    """
    Build and evaluate machine learning models.
    
    Args:
        df (DataFrame): Metabolite data
    """
    print("\n" + "=" * 80)
    print("PART 3: MACHINE LEARNING MODEL")
    print("=" * 80)
    
# Identify class column
    class_col = 'Class' if 'Class' in df.columns else 'class'
    
# Get numeric feature columns
    numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()
    feature_cols = [col for col in numeric_cols if col != class_col]
    
# Prepare features and target
    X = df[feature_cols].copy()
    y = df[class_col].copy()
    
# Handle any remaining NaN values
    X = X.fillna(X.mean())
    
    print(f"\nFeatures: {len(feature_cols)} metabolites")
    print(f"Target: {class_col} (Cachexic=1, Control=0)")
    print(f"Feature names (first 10): {[col[:20] for col in feature_cols[:10]]}")
    print(f"\nData shape: X={X.shape}, y={y.shape}")
    
# Check for class balance
    print(f"\nClass distribution in dataset:")
    print(f"  Cachexic (1): {(y == 1).sum()} samples")
    print(f"  Control (0): {(y == 0).sum()} samples")

# STEP ii for ML model: Train–test split    
# Split data
    print("\n" + "-" * 80)
    print("DATA PARTITIONING")
    print("-" * 80)
    
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, 
        test_size=0.3, 
        random_state=42, 
        stratify=y
    )
    
    print(f"\nTraining set: {len(X_train)} samples")
    print(f"  Cachexic: {(y_train == 1).sum()}")
    print(f"  Control: {(y_train == 0).sum()}")
    print(f"\nTest set: {len(X_test)} samples")
    print(f"  Cachexic: {(y_test == 1).sum()}")
    print(f"  Control: {(y_test == 0).sum()}")
    
# Step iii for ML model: Feature Standardization or Standardize features as it is termed
    print("\nStandardizing features...")
    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled = scaler.transform(X_test)
    print(" Features standardized (mean=0, std=1)")
  
# Step iv for ML model: Define models to compare and then, compare using cross-validation (AUC) 
    print("\n" + "-" * 80)
    print("MODEL COMPARISON (5-Fold Cross-Validation)")
    print("-" * 80)
    
    models = {
        'Logistic Regression': LogisticRegression(random_state=42, max_iter=1000),
        'Random Forest': RandomForestClassifier(n_estimators=100, random_state=42),
        'Gradient Boosting': GradientBoostingClassifier(n_estimators=100, random_state=42),
        'SVM': SVC(kernel='rbf', probability=True, random_state=42),
        'K-Nearest Neighbors': KNeighborsClassifier(n_neighbors=5)
    }
    
# Compare models
    print("\nEvaluating models using cross-validation...")
    results = {}
    
    for name, model in models.items():
        try:
            cv_scores = cross_val_score(
                model, 
                X_train_scaled, 
                y_train, 
                cv=5, 
                scoring='roc_auc'
            )
            
            mean_auc = cv_scores.mean()
            std_auc = cv_scores.std()
            
            results[name] = mean_auc
            
            print(f"\n{name}:")
            print(f"  Mean AUC: {mean_auc:.4f} (+/- {std_auc:.4f})")
            
        except Exception as e:
            print(f"\n{name}: Error during cross-validation: {e}")
            continue
    
    if not results:
        print("\nX ERROR: No models could be evaluated")
        return

# Step v for ML model: Selection of best model with the highest mean (cross-validated) AUC
    best_model_name = max(results, key=results.get)
    best_model = models[best_model_name]
    
    print("\n" + "-" * 80)
    print("BEST MODEL SELECTION")
    print("-" * 80)
    print(f"\n Best Model: {best_model_name}")
    print(f" Cross-validated AUC: {results[best_model_name]:.4f}")
    
# Step vi for ML model:: Train best model on the full training set
    print("\n" + "-" * 80)
    print("MODEL TRAINING")
    print("-" * 80)
    
    best_model.fit(X_train_scaled, y_train)
    print(f"\n {best_model_name} trained successfully on {len(X_train)} samples")

# Step vii for ML model:: : Evaluating the best model on the test set    
# .i.e. Make predictions (Predict class labels on the test data)
    y_pred = best_model.predict(X_test_scaled)
    
# Predict probabilities for the positive class (Cachexic = 1).
    y_pred_proba = best_model.predict_proba(X_test_scaled)[:, 1]
    
# Evaluate model
    print("\n" + "-" * 80)
    print("MODEL EVALUATION ON TEST SET")
    print("-" * 80)
    
    accuracy = accuracy_score(y_test, y_pred)
    print(f"\n Accuracy: {accuracy:.4f} ({accuracy*100:.2f}%)")

# Print a detailed classification report:
    print("\n Classification Report:")
    print("-" * 60)
    print(classification_report(
        y_test, 
        y_pred, 
        target_names=['Control (0)', 'Cachexic (1)'],
        digits=4
    ))

# Step viii for ML model:: ROC curve and AUC analysis   
    
# ROC Curve and AUC (.i.e. area under the ROC curve)
    print("-" * 80)
    print("ROC CURVE ANALYSIS")
    print("-" * 80)
    
    fpr, tpr, thresholds = roc_curve(y_test, y_pred_proba)
    roc_auc = auc(fpr, tpr)
    
    plt.figure(figsize=(10, 8))
    
# Plot ROC curve (.i.e. TPR vs FPR)
    plt.plot(
        fpr, tpr, 
        color='darkorange', 
        lw=3, 
        label=f'ROC curve (AUC = {roc_auc:.4f})'
    )
    
# Plot diagonal reference line (this represents random guessing)
    
    plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--', 
            label='Random Classifier (AUC = 0.50)')
    
# Necessary to Mark optimal threshold or point on the ROC curve.
    optimal_idx = np.argmax(tpr - fpr)
    optimal_threshold = thresholds[optimal_idx]
    plt.plot(fpr[optimal_idx], tpr[optimal_idx], 'ro', markersize=10, 
            label=f'Optimal Threshold = {optimal_threshold:.3f}')
    
    plt.xlim([0.0, 1.0])
    plt.ylim([0.0, 1.05])
    plt.xlabel('False Positive Rate', fontsize=12, fontweight='bold')
    plt.ylabel('True Positive Rate', fontsize=12, fontweight='bold')
    plt.title(f'ROC Curve - {best_model_name}', fontsize=14, fontweight='bold')
    plt.legend(loc="lower right", fontsize=11)
    plt.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.savefig('roc_curve.png', dpi=300, bbox_inches='tight')
    print(f"\n ROC curve saved: roc_curve.png")
    plt.show()
    
    print(f"\n AUC Score: {roc_auc:.4f}")
    
# Step ix for ML model:: Feature importance (for tree-based models)  
    
# to show how much each feature contributes.)
    if hasattr(best_model, 'feature_importances_'):
        print("\n" + "-" * 80)
        print("FEATURE IMPORTANCE ANALYSIS")
        print("-" * 80)
        
        feature_importance = pd.DataFrame({
            'Feature': [col[:30] for col in feature_cols],  # Truncatation of long names
            'Importance': best_model.feature_importances_
        }).sort_values('Importance', ascending=False)
        
        print("\n Top 15 Most Important Metabolites:")
        print("-" * 60)
        for idx, row in feature_importance.head(15).iterrows():
            print(f"  {row['Feature']:<30} : {row['Importance']:.4f}")
        
# Plotting the top 15 most important features
        plt.figure(figsize=(10, 8))
        top_features = feature_importance.head(15)
        
        colors = plt.cm.viridis(np.linspace(0.3, 0.9, len(top_features)))
        
        plt.barh(range(len(top_features)), top_features['Importance'], 
                color=colors, edgecolor='black', linewidth=0.5)
        plt.yticks(range(len(top_features)), top_features['Feature'], fontsize=9)
        plt.xlabel('Importance Score', fontsize=12, fontweight='bold')
        plt.title('Top 15 Feature Importances', fontsize=14, fontweight='bold')
        plt.gca().invert_yaxis()
        plt.grid(axis='x', alpha=0.3, linestyle='--')
        
        plt.tight_layout()
        plt.savefig('feature_importance.png', dpi=300, bbox_inches='tight')
        print("\n Feature importance plot saved: feature_importance.png")
        plt.show()
    
    
# Step X: Overall Summary
    print("\n" + "=" * 80)
    print("MODEL PERFORMANCE SUMMARY")
    print("=" * 80)
    print(f"\n Best Model: {best_model_name}")
    print(f" Test Accuracy: {accuracy*100:.2f}%")
    print(f" AUC Score: {roc_auc:.4f}")
    print(f" Model successfully trained and evaluated!")

# Running analysis
def main_analysis():
    """
    Main function to run complete analysis pipeline.
    """
    print("\n")
    print("=" * 80)
    print("  TASK 2.2: CANCER-ASSOCIATED SKELETAL MUSCLE WASTING ANALYSIS")
    print("=" * 80)
    print("\n")
    
# My actual file path
    excel_file = 'Downloads/Python_Final_assessment/metabolite 2.2.xlsx'
    
    print(f" Searching for file: {excel_file}\n")
    
# Part 1: Data Preparation
    df = load_and_prepare_data(excel_file)
    
    if df is None:
        print("\nX Cannot proceed without data. Please check file path.")
        return
    
# Part 2: Data Visualization
    try:
        visualize_data(df)
    except Exception as e:
        print(f"\n [Warning] during visualization: {e}")
        import traceback
        traceback.print_exc()
        print("\nContinuing with machine learning...")
    
# Part 3: Machine Learning
    try:
        build_machine_learning_model(df)
    except Exception as e:
        print(f"\nX Error during machine learning: {e}")
        import traceback
        traceback.print_exc()
        return

# In all Summary
    
    print("\n" + "=" * 80)
    print(" ANALYSIS COMPLETE - ALL TASKS FINISHED SUCCESSFULLY")
    print("=" * 80)
    print("\n Generated Files:")
    print("   metabolite_distributions.png")
    print("   class_comparison_boxplots.png")
    print("   correlation_heatmap.png")
    print("   pca_scatter_plot.png")
    print("   roc_curve.png")
    print("   feature_importance.png")
    print("\n" + "=" * 80)


# PRIMARY EXECUTION


if __name__ == "__main__":
    main_analysis()

```
